```python
#Data Manipulation and math
import random
import numpy as np
import pandas as pd

#System performance and Benchmarking
import psutil
import time
import json
import os
from datetime import datetime
```

Implementation of Linear Search 

- It sequentially searches the input array for the target value and returns whether it was found, its index, and the number of comparisons performed.

```python
class LinearSearch:

    def search(arr, target):

        comparisons = 0

        for index, value in enumerate(arr):

            comparisons += 1

            if value == target:
                return {
                    "found": True,
                    "index": index,
                    "comparisons": comparisons
                }

        return {
            "found": False,
            "index": -1,
            "comparisons": comparisons
        }
```

Generates randomized input arrays of the specified sizes to be used for benchmarking the linear search algorithm.

```python
def generate_array(size):

    arr = list(range(size))
    random.shuffle(arr)

    return arr
```

Executes the linear search algorithm across different input sizes and trials while recording:
- runtime
- CPU time
- throughput
- memory usage

and the number of comparisons performed.

```python
def benchmark_linear_search(array_sizes, trials=10):

    process = psutil.Process(os.getpid())
    results = []

    for size in array_sizes:

        print(f"Benchmarking array size {size}...")

        for trial in range(1, trials + 1):

            arr = generate_array(size)

            # Worst case
            target = -1

            cpu_start = time.process_time()
            wall_start = time.perf_counter()

            output = LinearSearch.search(arr, target)

            wall_end = time.perf_counter()
            cpu_end = time.process_time()

            runtime = wall_end - wall_start
            cpu_time = cpu_end - cpu_start

            throughput = size / runtime

            peak_ram = process.memory_info().rss / (1024 * 1024)

            results.append({

                "Array Size": size,
                "Trial": trial,
                "Runtime (s)": runtime,
                "CPU Time (s)": cpu_time,
                "Throughput (elements/s)": throughput,
                "Peak RAM (MB)": peak_ram,
                "Comparisons": output["comparisons"],
                "Found": output["found"]

            })

    return results
```

Defines the input array sizes and executes the benchmarking function over multiple trials, storing the performance metrics.

```python
array_sizes = [8, 16, 24, 32]

results = benchmark_linear_search(
    array_sizes=array_sizes,
    trials=10
)
```

Organizes the collected benchmark results into a DataFrame.

```python
df = pd.DataFrame(results)

display(df)
```

Groups the benchmark results by input size and computes the mean and standard deviation of the recorded performance metrics across all trials.

```python
summary_df = (
    df.groupby("Array Size")
      .agg({
          "Runtime (s)": ["mean", "std"],
          "CPU Time (s)": ["mean", "std"],
          "Throughput (elements/s)": ["mean", "std"],
          "Peak RAM (MB)": ["mean", "std"],
          "Comparisons": ["mean", "std"]
      })
)

summary_df
```

Exports the detailed benchmark results to linear_search_results.csv and the aggregated summary statistics to linear_search_summary.csv. 

The detailed results include the array size, trial number, runtime, CPU time, throughput, peak RAM usage, number of comparisons, and search result, while the summary file contains the mean and standard deviation of the recorded performance metrics for each input size.

```python
df.to_csv("linear_search_results.csv", index=False)

summary_df.to_csv("linear_search_summary.csv")
```
